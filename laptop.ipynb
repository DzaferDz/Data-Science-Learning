```python
import pandas as pd

df = pd.read_csv('laptop_price.csv', encoding='cp1252')
# oder
# df = pd.read_csv('laptop_price.csv', encoding='latin1')
```

```python
df.head(1)
```

```python
# Opcija A: Rječnik za preimenovanje specifičnih kolona
df = df.rename(columns={"Company": "Unternehmen", "Product": "Produkt", "TypeName": "Name", "ScreenResolution": "Bildschirmauflösung", "Cpu": "CPU", "Ram": "RAM", "Memory": "Speicher", "Gpu": "GPU", "OpSys": "OS", "Weight": "Gewicht", "Price_euros": "Preis(EUR)"})
```

```python
df.head(1)
```

```python
df=df.rename(columns = {"Inches": "Zoll"})
```

```python
df.info()^
```

```python
df['RAM'] = df['RAM'].str.replace('GB', '').astype(int)
```

```python
df['SpeicherAnzahl'] = df['Speicher'].str.extract('(\d+)').astype(int)
```

```python
df['Hat_SSD'] = df['Speicher'].str.contains('SSD').astype(int)

# Prüfen, ob HDD im Text steht
df['Hat_HDD'] = df['Speicher'].str.contains('HDD').astype(int)
```

```python
print(df['Bildschirmauflösung'].head(3))

# Neue Spalte erstellen: Hat der Laptop ein IPS-Panel? (True / False)
df['Ips_Panel'] = df['Bildschirmauflösung'].str.contains('IPS Panel').astype(int)

# Neue Spalte erstellen: Ist es ein Touchscreen? (True / False)
df['Touchscreen'] = df['Bildschirmauflösung'].str.contains('Touchscreen').astype(int)

# Zeigen, was neu entstanden ist
df[['Bildschirmauflösung', 'Ips_Panel', 'Touchscreen']].head()
```

```python
df['Gewicht'] = df['Gewicht'].str.replace('kg', '').astype(float)

# Überprüfen
print(df['Gewicht'].dtype)
df['Gewicht'].head()
```

```python
preis_pro_marke = df.groupby('Unternehmen')['Preis(EUR)'].mean().sort_values(ascending=False)
print("Durchschnittpreis nach Hersteller: ")
print(preis_pro_marke.round(2))
```
