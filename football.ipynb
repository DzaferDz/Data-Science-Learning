```python
import pandas as pd

root = "https://www.football-data.co.uk/mmz4281/"
leagues = ['E0', 'E1']
frames = []

for league in leagues:
    # range(15, 27) ide od 15 do 26 uključujući 26 (što daje 2627 / sezonu 26/27)
    for season in range(15, 27):
        s_start = f"{season:02d}"
        s_end = f"{(season + 1) % 100:02d}"
        season_str = f"{s_start}{s_end}"     # npr. "1516", "2627"
        season_fmt = f"{s_start}/{s_end}"     # npr. "15/16", "26/27"
        
        file_path = f"{root}{season_str}/{league}.csv"
        
        try:
            df = pd.read_csv(file_path)
            
            # Dodavanje 'Season' kolone na drugo mjesto (indeks 1)
            df.insert(1, 'Season', season_fmt)
            
            frames.append(df)
            print(f"Uspješno učitano: {season_fmt} - {league}")
        except Exception as e:
            print(f"Greška pri učitavanju {season_fmt} za {league}: {e}")

# Spajanje svih tabela u jednu veliku
full_df = pd.concat(frames, ignore_index=True)
```
